```python
try:
    age = int(input("Enter the age:"))
    if(age<18):
        raise ValueError
    else:
        print("the age is valid")
        
except ValueError:
    print("The age is not valid")
```

```python
try:
    num=int(input("Enter a number"))
    print(f"the entered number is {num}")
except ValueError as ex:
    print("Exception:", ex)
```

```python
num=int(input("Enter a number to find if its prime or not:"))
counter=0
for i in range(2,num):
    if num%i==0:
        counter+=1
        break
if counter>0:
    print(f"{num} is not prime")
else:
    print(f"{num} is prime")
```

```python
try:
    num1,num2=eval(input("Enter two number separated by commas:"))
    result=num1/num2
    print(f"The result is {result}")
except SyntaxError:
    print("Comma is missing, please enter a comma between the numbers like this: 1,2")
except ZeroDivisionError:
    print("Division by zero is error!!")
except:
    print("Wrong input.")
else:
    print("No exceptions")
finally:
    print("This will execute no matter what")
```
